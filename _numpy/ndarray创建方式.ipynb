```python
# 从0到10的等差数列,5个数,endpoint为true包含最后一个数
np.linspace(start=0,stop=10,num=5,endpoint=True)

# 以2为底数,2的2次~4的2次,4个数
np.logspace(2,4,num=4,base=2)
```
